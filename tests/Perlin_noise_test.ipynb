```python
import numpy as np
import matplotlib.pyplot as plt
from sequence.kernel.timeline import Timeline
from simtwo.core.JitterNode import RxNodeWithJitter

rng = np.random.default_rng(123)
timeline = Timeline()

rx = RxNodeWithJitter(
    "B",
    timeline,
    rng=rng,
    jitter_std_ps=50_000,
    jitter_method="default",
    perlin_step=0.05,
)

jitter_ps = np.array([rx._sample_jitter_ps() for _ in range(1000)])

plt.plot(jitter_ps)
plt.xlabel("Sample")
plt.ylabel("Jitter (ps)")
plt.show()
```

```python
timeline = Timeline()
rx = RxNodeWithJitter(
    "B",
    timeline,
    rng=rng,
    jitter_std_ps=50_000,
    jitter_method="perlin",
    perlin_step=0.25,
)

jitter_ps = np.array([rx._sample_jitter_ps() for _ in range(1000)])

plt.plot(jitter_ps)
plt.xlabel("Sample")
plt.ylabel("Jitter (ps)")
plt.show()
```
